```python
import uproot
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
filename = "/exp/sbnd/data/users/castalyf/PMT_timing/LED_data/wvfm_ana_undeco-EventBuilder5_art1_run19410_7_20250924T165530.root"
# filename = "/exp/sbnd/data/users/castalyf/PMT_timing/LED_data/wvfm_ana_undeco-EventBuilder6_art1_run19440_2_20250924T220813.root"
file = uproot.open(filename)
file.keys()
```

```python
# Get all keys except the directory itself
all_keys = [key for key in file.keys() if key != 'flashana;1']

# Threshold for PMT response detection
threshold = 2000

print(f"Scanning {len(all_keys)} channels for PMT responses...")
print(f"Threshold: max(waveform) - min(waveform) > {threshold}")
print("-" * 60)

channels_with_response = []

for key in all_keys:
    try:
        # Read the waveform data
        waveform = file[key].values()
        
        # Calculate the range (max - min)
        wf_max = np.max(waveform)
        wf_min = np.min(waveform)
        wf_range = wf_max - wf_min
        
        # Check if it exceeds threshold
        if wf_range > threshold:
            # Extract the channel name (remove the version number)
            channel_name = key.split(';')[0].replace('flashana/', '')
            channels_with_response.append({
                'key': key,
                'name': channel_name,
                'range': wf_range,
                'max': wf_max,
                'min': wf_min
            })
            
    except Exception as e:
        print(f"Error processing {key}: {e}")
        continue

print(f"\nFound {len(channels_with_response)} channels with potential PMT response:\n")

# Sort by range (largest first) and print
channels_with_response.sort(key=lambda x: x['range'], reverse=True)

for ch in channels_with_response:
    print(f"{ch['name']}")
    print(f"  Range: {ch['range']:.1f} (max: {ch['max']:.1f}, min: {ch['min']:.1f})")
    print()

# # Optional: Save to file
# with open('pmt_channels_with_response.txt', 'w') as f:
#     f.write(f"Channels with PMT response (threshold: {threshold})\n")
#     f.write("=" * 60 + "\n\n")
#     for ch in channels_with_response:
#         f.write(f"{ch['name']}\n")
#         f.write(f"  Range: {ch['range']:.1f} (max: {ch['max']:.1f}, min: {ch['min']:.1f})\n\n")

# print(f"Results saved to 'pmt_channels_with_response.txt'")
```
